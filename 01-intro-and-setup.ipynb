## What is Scala?

Scala is a **statically typed, hybrid functional and object-oriented** language that runs on the **Java Virtual Machine** (JVM).

It was created by Martin Odersky and his team at EPFL in 2004 with one clear goal: take the engineering pragmatism of Java and fuse it with the expressive power of functional programming. The name is a contraction of *scalable language* — code is meant to grow from quick scripts into large systems without rewriting.

Four traits that define Scala in one breath:

- **Statically typed but inference-heavy** — the compiler catches type errors, yet you rarely need to write the types yourself.
- **Functions are values** — pass them around, return them, store them in collections, just like integers.
- **Immutable by default** — `val` (a value that cannot be reassigned) is preferred over `var`.
- **Runs on the JVM** — it sees every Java library on Maven Central as part of its standard library.

## Why Scala?

Three reasons people pick Scala in 2026:

| Reason | What it unlocks |
|---|---|
| **Spark and big-data** | Apache Spark itself is written in Scala. The native Spark API surface, internals, and tuning knobs are all expressed in Scala first. |
| **Type-safe domain modeling** | Algebraic data types, pattern matching, and a powerful type system let you encode invariants the compiler will enforce. |
| **JVM ecosystem with FP ergonomics** | Every Java library, every Kafka client, every Hadoop tool — but with `map`, `flatMap`, and pattern matching as first-class citizens. |

For this learning track the **Spark connection** is the centre of gravity. Every notebook from here on will, where natural, point out the Scala feature being used inside Spark's own codebase.

## Scala 3, not Scala 2

This track uses **Scala 3** — the version line that started in 2021, codename *Dotty*.

Scala 3 is a deep redesign: same JVM target, same flavour, but a cleaner type system and a much friendlier surface syntax. Two highlights you will feel immediately:

- **Significant indentation** — code blocks can be expressed with indentation, the way Python does, instead of curly braces. Braces still work; you choose per file.
- **`given` and `using` replace `implicit`** — Scala 2's most powerful and most confusing feature was rebuilt with clearer keywords. You will meet these in notebook 11.

If you stumble onto a Stack Overflow answer using `implicit val foo`, that's Scala 2 vintage. The shape of the idea is unchanged; only the keyword differs.

## The JVM context

When you write Scala, the compiler `scalac` turns your source code into **JVM bytecode** — the same instruction set Java compiles to. The JVM then loads that bytecode and executes it.

```text
  your_file.scala
        |
        v
    [ scalac ]        <- Scala 3 compiler
        |
        v
   your_file.class    <- JVM bytecode
        |
        v
      [ JVM ]         <- loads bytecode, GC, JIT-compiles hot paths
        |
        v
      output
```

Two consequences worth absorbing now:

1. **Startup is not instant.** The JVM itself takes a moment to boot. For tiny scripts this feels slow; for long-running services it disappears.
2. **You inherit the JVM's strengths.** A mature garbage collector, the JIT compiler that optimises hot code paths at runtime, and decades of profiling tools.

## Three ways to run Scala

Scala gives you three on-ramps. Pick the one that matches the size of what you are doing.

```text
  +-------------+   +---------------+   +------------------+
  |    REPL     |   |   scala-cli   |   |       sbt        |
  |  (one-off)  |   |  (one file)   |   |  (real project)  |
  +-------------+   +---------------+   +------------------+
   live eval        run a .scala     many files, deps,
   no file          like a script    tests, build config
```

- **REPL** — for *what does this expression do?* The cheapest feedback loop.
- **`scala-cli`** — for a single file you want to keep around. Add dependencies via a comment at the top of the file.
- **`sbt`** — for anything that has more than one file or that you want to build, package, and test seriously. This is what Spark itself uses.

## Hello, REPL

Open the REPL with the `scala` command. You get a `scala>` prompt that evaluates one expression at a time.

````scala
scala> println("hello, scala 3")
hello, scala 3

scala> val name = "ganesh"
val name: String = ganesh

scala> s"hello, $name"
val res0: String = hello, ganesh

scala> :quit
````

Three details to notice in that transcript:

- `val name = "ganesh"` binds a value. The REPL echoes back the inferred type — `String`.
- `s"hello, $name"` is a **string interpolator**. The `s` prefix lets `$name` and `${...}` be spliced into the string at compile time.
- `:quit` exits. (`:help` shows all REPL commands.)

## Hello, `scala-cli`

For anything longer than two lines, put it in a file and run it with `scala-cli`. Create `hello.scala`:

````scala
@main def hello(): Unit =
  println("hello from scala-cli")
````

Run it from the same directory:

````scala
%%bash
scala-cli run hello.scala
# hello from scala-cli
````

Two new shapes appeared:

- `@main def hello()` — the `@main` annotation marks an entry point. Whatever method carries it becomes the program's `main`. No more `object Main extends App` boilerplate from Scala 2.
- `: Unit =` — `Unit` is Scala's equivalent of `void`. It is a real type with exactly one value, written `()`.

## Hello, `sbt`

For a real project, **sbt** (the *simple build tool*) gives you a directory layout, dependency management, incremental compilation, and a test runner. You will meet it properly in notebook 15 when this track bridges to Spark. For now, the one-command scaffold is enough:

````scala
%%bash
sbt new scala/scala3.g8
````

That pulls a template, asks for a project name, and creates a directory with the standard layout:

```text
  myproject/
  +-- build.sbt              <- project config: name, version, scala version, deps
  +-- project/
  |    +-- build.properties
  +-- src/
       +-- main/scala/       <- production code goes here
       +-- test/scala/       <- tests go here
```

Run `sbt run` inside the directory to compile and execute.

## A note on syntax style

Scala 3 supports two equivalent surface syntaxes for code blocks. The same method, both ways:

````scala
// indentation style — Scala 3 idiomatic
def greet(name: String): String =
  val greeting = s"hello, $name"
  greeting.toUpperCase

// brace style — still valid, common in older code and in Spark
def greet(name: String): String = {
  val greeting = s"hello, $name"
  greeting.toUpperCase
}
````

This track uses **indentation style** by default — it matches the Scala 3 idiomatic direction and removes visual noise. When you read Spark's source, you will see brace style, because Spark itself still targets Scala 2.13. Both shapes parse the same tree; pick whichever reads better in the file you are editing.
